# Verify the DoCIT results

Run all cells in a CPU runtime. In Colab, upload `DoCIT_Anonymous.zip` when the first cell prompts you. In local Jupyter, open this notebook from the extracted package directory.

The checks rerun the three-project Phase 1 preview and recompute the archived iTrust pilot analysis. The iTrust pilot notebook can rerun the CUDA model inference.

```python
from pathlib import Path
from tempfile import mkdtemp
from zipfile import ZipFile

package = Path.cwd()
if not (package / "verify_results.py").is_file():
    archive = package / "DoCIT_Anonymous.zip"
    if not archive.is_file():
        try:
            from google.colab import files
        except ImportError as error:
            raise FileNotFoundError("Put DoCIT_Anonymous.zip in the notebook directory.") from error
        print("Upload DoCIT_Anonymous.zip.")
        files.upload()
    if not archive.is_file():
        raise FileNotFoundError("Expected DoCIT_Anonymous.zip in the current directory.")
    stage = Path(mkdtemp(prefix="docit_bundle_"))
    with ZipFile(archive) as bundle:
        bundle.extractall(stage)
    package = stage / "DoCIT_Verification"

if not (package / "verify_results.py").is_file():
    raise FileNotFoundError("The verification script is missing from the extracted package.")
print("Verification package ready.")
```

```python
import os
import subprocess
import sys

environment = Path(mkdtemp(prefix="docit_python_"))
subprocess.run([sys.executable, "-m", "venv", str(environment)], check=True)
python = environment / ("Scripts/python.exe" if os.name == "nt" else "bin/python")
subprocess.run([
    str(python), "-m", "pip", "install", "-q", "-r", str(package / "requirements.txt")
], check=True)
```

```python
subprocess.run([str(python), str(package / "verify_results.py")], check=True)
```
